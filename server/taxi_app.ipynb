```python
TAXI_WAIT_URL = "https://raw.githubusercontent.com/adi1zhexen0v/mnu-datasets/main/taxi_wait_extended.csv"
```

```python
import numpy as np                                    
import pandas as pd                                   
import matplotlib.pyplot as plt                       
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv(TAXI_WAIT_URL)
df["weather"] = df["weather"].map({
    "clear": 0, 
    "rain": 1, 
    "snow": 2}
)

X = df[["driver_distance_km", "drivers_nearby", "hour", "weather"]]
y = df["wait_minutes"]

model = LinearRegression()
model.fit(X, y)

print(model.coef_, model.intercept_)
```

```python
import joblib

joblib.dump(model, "taxi_app_model.joblib")
```
